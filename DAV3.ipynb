```python
import pandas as pd

# Assuming the uploaded file is 'research_student (1) (1).xlsx'
file_name = 'research_student (1) (1).xlsx'
df = pd.read_excel(file_name)

display(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Drop rows where 'Marks[10th]' is NaN to ensure the histogram plots correctly
df_marks_10th = df.dropna(subset=['Marks[10th]'])

plt.figure(figsize=(10, 6))
sns.histplot(df_marks_10th['Marks[10th]'], kde=True, bins=15)
plt.title('Distribution of 10th Grade Marks')
plt.xlabel('Marks[10th]')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(y=df['Marks[10th]'].dropna())
plt.title('Box Plot of 10th Grade Marks')
plt.ylabel('Marks[10th]')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(y=df['Marks[12th]'].dropna())
plt.title('Box Plot of 12th Grade Marks')
plt.ylabel('Marks[12th]')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(y=df['GPA 1'].dropna())
plt.title('Box Plot of GPA 1')
plt.ylabel('GPA 1')
plt.show()
```

```python
plt.figure(figsize=(12, 7))
sns.boxplot(x='Branch', y='GPA 1', data=df.dropna(subset=['Branch', 'GPA 1']), palette='viridis')
plt.title('GPA 1 Distribution Across Branches')
plt.xlabel('Branch')
plt.ylabel('GPA 1')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Drop rows where 'Gender' is NaN to ensure the countplot plots correctly
df_gender = df.dropna(subset=['Gender'])

plt.figure(figsize=(8, 6))
sns.countplot(x='Gender', data=df_gender, palette='viridis')
plt.title('Count of Students by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(10, 7))
sns.scatterplot(x='Marks[10th]', y='Marks[12th]', data=df.dropna(subset=['Marks[10th]', 'Marks[12th]']))
plt.title('10th Grade Marks vs. 12th Grade Marks')
plt.xlabel('Marks[10th]')
plt.ylabel('Marks[12th]')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
# Drop rows where 'Category' is NaN to ensure the countplot plots correctly
df_category = df.dropna(subset=['Category'])

plt.figure(figsize=(10, 6))
sns.countplot(x='Category', data=df_category, palette='magma', order=df_category['Category'].value_counts().index)
plt.title('Count of Students by Category')
plt.xlabel('Category')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(16, 12))
sns.heatmap(df.select_dtypes(include=['number']).corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```
